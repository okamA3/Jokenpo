```python
print("-----Seja bem vindo ao Jogo de Par ou Impar-----\n")
jogador1 = int(input("Escolha par ou impar: 0 = Par e 1 = Impar: \n"))
jogador2 = ""
par = 0
impar = 1

if jogador1 == 0:
  impar = "jogador2"
  par = "jogador1"
  print("Jogador 1 escolheu par, Jogador 2 é impar")
elif jogador1 == 1:
  par = "jogador2"
  impar = "jogador1"
  print("Jogador 1 escolheu impar, Jogador 2 é par")
else:
  print("Opção inválida")

jogador2nmr = int(input("Digite um número inteiro qualquer para o Jogador 2: "))
jogador1nmr = int(input("Digite um número inteiro qualquer para o Jogador 1: \n"))

soma = jogador1nmr + jogador2nmr

if soma % 2 == 0:
  print(f"O número é par, o {par} venceu!")

else:
  print(f"O número é impar, o {impar} venceu!")

```
